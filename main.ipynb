```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

df = pd.read_csv("data-final.csv", delimiter="\t")
df.head()

```python
reverse_questions = [
    'EXT2', 'EXT4', 'EXT6', 'EXT8', 'EXT10',
    'EST1', 'EST3', 'EST5', 'EST6', 'EST7', 'EST8', 'EST9', 'EST10',
    'AGR1', 'AGR3', 'AGR5', 'AGR7',
    'CSN2', 'CSN4', 'CSN6', 'CSN8',
    'OPN2', 'OPN4', 'OPN6'
]

#Inversion formula
df[reverse_questions] = 6 - df[reverse_questions]

#Adding aggregate columns
df['EXT'] = df[['EXT1','EXT2','EXT3','EXT4','EXT5',
                 'EXT6','EXT7','EXT8','EXT9','EXT10']].sum(axis=1)
df['EST'] = df[['EST1','EST2','EST3','EST4','EST5',
                 'EST6','EST7','EST8','EST9','EST10']].sum(axis=1)
df['AGR'] = df[['AGR1','AGR2','AGR3','AGR4','AGR5',
                 'AGR6','AGR7','AGR8','AGR9','AGR10']].sum(axis=1)
df['CSN'] = df[['CSN1','CSN2','CSN3','CSN4','CSN5',
                 'CSN6','CSN7','CSN8','CSN9','CSN10']].sum(axis=1)
df['OPN'] = df[['OPN1','OPN2','OPN3','OPN4','OPN5',
                 'OPN6','OPN7','OPN8','OPN9','OPN10']].sum(axis=1)
```
