```python
def viz_questions(traitgroup, questions, color):
    plt.figure(figsize=(40,60))
    for i in range(1,11):
        plt.subplot(10,5,i)
        plt.hist(df[traitgroup[i-1]], bins=14, color=color, alpha=.5)
        plt.title(questions[traitgroup[i-1]], fontsize=18)
```
